```python
import pandas as pd
import numpy as np

path = "US Dollar Index Historical Data.csv"  
df = pd.read_csv(path)

print(df.shape)
df.head()
```

```python
df.info()
df.isna().mean().sort_values(ascending=False)
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%m/%d/%Y", errors="raise")

df["Date"] = df["Date"].dt.to_period("M").dt.to_timestamp()  # YYYY-MM-01

df = df.sort_values("Date").reset_index(drop=True)

df[["Date"]].head(), df[["Date"]].tail()
```

```python
df["Change %"] = (
    df["Change %"]
    .astype(str)
    .str.replace("%", "", regex=False)
    .str.replace(",", ".", regex=False)
)

df["Change %"] = pd.to_numeric(df["Change %"], errors="coerce")

df["Change %"].describe()
```

```python
df = df.drop(columns=["Vol."])
df.head()
```

```python
num_cols = [c for c in df.columns if c != "Date"]
for c in num_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

# Basit tutarlılık kontrolü
bad = df[(df["High"] < df["Low"]) | (df["High"] < df["Price"]) | (df["Low"] > df["Price"])]
print("# of inconsistent row:", len(bad))

# Date unique mi?
print("# of Duplicate date:", df["Date"].duplicated().sum())

# Kaydet
df.to_csv("US_Dollar_Index_cleaned.csv", index=False)
df.tail()
```
